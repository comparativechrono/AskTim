# AskTim! in a notebook

AskTim! is an AI helper for bioinformatics that keeps your data out of the AI. In this notebook you will:

1. ask it a question about the tools in this notebook,
2. let it explain an error,
3. check exactly what it sends to the AI, and
4. let its agent do a small analysis, approving each step.

The data here are made up, so nothing real is at stake, but AskTim! treats them as if they were real: the AI sees their *shape* (file names, sizes, column names) and never their contents.

## 3. Some example data

This makes a small FASTQ file of (made-up) reads and a table of (made-up) gene counts.

```python
import random
random.seed(1)
with open("reads.fastq", "w") as f:
    for i in range(500):
        n = random.randint(60, 150)
        seq = "".join(random.choice("ACGT") for _ in range(n))
        f.write("@read%d\n%s\n+\n%s\n" % (i, seq, "F" * n))
with open("counts.csv", "w") as f:
    f.write("gene,control_1,control_2,treated_1,treated_2\n")
    for g in range(200):
        f.write("gene%03d,%d,%d,%d,%d\n" % (g, *[random.randint(0, 2000) for _ in range(4)]))
print("Wrote reads.fastq and counts.csv")
```

## 4. Ask a question

Start a cell with `AskTim!` and a question in quotes. AskTim! knows which packages this notebook has, so its answer uses those.

```python
AskTim! "How do I find the number of reads in reads.fastq and their mean length, in Python?"
```

`AskTim! insert 1` would put the first code block of that answer into a new cell for you (in Colab: a box with a Copy button).

## 6. What did AskTim! send?

Using AI responsibly includes knowing what you share. `AskTim! show` prints exactly what would be sent with a question, and sends nothing. Look for your data: the file names and sizes are there, the reads are not.

```python
AskTim! show "What is the average count of the control samples?"
```

## 7. Let the agent do it

The agent plans the work and carries it out step by step: it can run Python in this notebook, run programs, write files and add cells. Before each step you see what it wants to do and why, and you answer in the box that appears (Enter or `y` runs it; `n` refuses it, and you can say why). It can only use what is installed, cannot install anything, and sees what its steps print with the data lines withheld.

```python
AskTim! Agent "Count the reads in reads.fastq, plot a histogram of their lengths and save it as lengths.png"
```

The agent left a record of everything it did in `asktim-runs/` (`RUN.md`), including what the AI was shown at each step.
